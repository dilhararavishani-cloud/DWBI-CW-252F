## STEP 2 — Design a Data Warehouse

Based on the Customer 360 data, we will design a **Star Schema**. This model separates business process data (measures) into fact tables and descriptive attributes into dimension tables, optimizing for query performance and ease of understanding.

### Star Schema Design

**Fact Tables** will store quantitative data about business events, linked by foreign keys to dimensions. **Dimension Tables** will store qualitative, descriptive data, providing context to the facts.

### Dimension Tables

Dimension tables provide descriptive context for the numerical facts. They are typically denormalized for query performance.

1.  **`dim_customer`**
    *   **Purpose:** Stores static and demographic information about each customer.
    *   **Justification:** Essential for analyzing customer behavior, segmenting customers by age, gender, or other attributes, and understanding customer demographics in relation to sales, support, and clickstream data.
    *   **Business Key:** `customer_id`
    *   **Proposed Columns:** `customer_key` (Surrogate Key), `customer_id` (Business Key), `name` (customer name), `email`, `age`, `gender`.

2.  **`dim_product`**
    *   **Purpose:** Contains descriptive attributes for each product.
    *   **Justification:** Enables analysis of sales performance, customer preferences, and support issues based on product categories, names, or other product-specific details.
    *   **Business Key:** `product_id`
    *   **Proposed Columns:** `product_key` (Surrogate Key), `product_id` (Business Key), `name` (product name), `category`.

3.  **`dim_date`**
    *   **Purpose:** Provides a rich set of date-related attributes for time-based analysis.
    *   **Justification:** Critical for analyzing trends, seasonality, and time-specific patterns across orders, support tickets, and clickstream events. It centralizes all date information.
    *   **Business Keys:** Derived from `order_date` (from `orders`) and `timestamp` (from `clickstream`).
    *   **Proposed Columns:** `date_key` (Surrogate Key, e.g., YYYYMMDD), `full_date`, `day`, `day_name`, `week`, `month`, `month_name`, `quarter`, `year`.

4.  **`dim_support_status`**
    *   **Purpose:** Stores distinct statuses for support tickets.
    *   **Justification:** Allows for standardized reporting and analysis of ticket resolution workflows, common statuses, and the efficiency of the support process.
    *   **Business Key:** `status`
    *   **Proposed Columns:** `support_status_key` (Surrogate Key), `status` (e.g., 'Open', 'Closed').

5.  **`dim_page`**
    *   **Purpose:** Stores distinct page URLs or names visited by customers.
    *   **Justification:** Facilitates analysis of user navigation, popular content, and engagement patterns on the website, allowing insights into website design and content effectiveness.
    *   **Business Key:** `page_visited`
    *   **Proposed Columns:** `page_key` (Surrogate Key), `page_name`.


### Fact Tables

Fact tables contain the quantitative measures and foreign keys linking to dimension tables, representing specific business events.

1.  **`fact_orders`**
    *   **Purpose:** Records each individual order event and its associated numerical measures.
    *   **Justification:** This is the core for sales analysis, revenue tracking, and understanding product demand. Each row represents a specific order, detailing what was bought, when, by whom, and for how much.
    *   **Grain:** One row per order item.
    *   **Proposed Columns:** `order_id` (Degenerate Dimension/Business Key), `customer_key` (FK to `dim_customer`), `product_key` (FK to `dim_product`), `date_key` (FK to `dim_date` based on `order_date`), `quantity`, `price` (unit price at the time of order), `total_amount`.

2.  **`fact_support_tickets`**
    *   **Purpose:** Records details for each support ticket event.
    *   **Justification:** Enables analysis of customer service performance, identifying common issues, and tracking resolution times (if date fields were available).
    *   **Grain:** One row per support ticket.
    *   **Proposed Columns:** `ticket_id` (Degenerate Dimension/Business Key), `customer_key` (FK to `dim_customer`), `support_status_key` (FK to `dim_support_status`), `issue`.
    *   **Note on Date:** The current `support` DataFrame lacks a specific date column for ticket creation. In a real-world scenario, a `ticket_creation_date` would be crucial for linking to `dim_date`.

3.  **`fact_clickstream`**
    *   **Purpose:** Records each individual page visit event by a customer.
    *   **Justification:** Provides granular data for analyzing user behavior on the website, including navigation paths, page popularity, and engagement metrics.
    *   **Grain:** One row per page visit event within a session.
    *   **Proposed Columns:** `session_id` (Degenerate Dimension/Business Key), `customer_key` (FK to `dim_customer`), `date_key` (FK to `dim_date` based on `timestamp`), `page_key` (FK to `dim_page`).


## STEP 3 — Create Surrogate Keys

Surrogate keys are system-generated, simple integer keys used in data warehousing to uniquely identify records in dimension tables, independent of the source system's business keys. They provide performance benefits, flexibility in handling source system changes, and ensure uniqueness.

*   **Business Key:** An identifier for a record from the source system (e.g., `customer_id`, `product_id`). These can sometimes change or be reused, and might be composite.
*   **Surrogate Key:** A new, simple, integer primary key generated specifically for the data warehouse. It is stable, non-intelligent, and ensures referential integrity within the warehouse.

I will now create these surrogate keys for our dimension tables, keeping the original business keys for traceability.

```python
# Create dim_customer with surrogate key
df_dim_customer = customers[['customer_id', 'name', 'email', 'age', 'gender']].copy()
df_dim_customer = df_dim_customer.drop_duplicates(subset=['customer_id'])
df_dim_customer['customer_key'] = range(1, len(df_dim_customer) + 1)
df_dim_customer = df_dim_customer[['customer_key', 'customer_id', 'name', 'email', 'age', 'gender']]

# Create dim_product with surrogate key
df_dim_product = products[['product_id', 'name', 'category']].copy()
df_dim_product = df_dim_product.drop_duplicates(subset=['product_id'])
df_dim_product['product_key'] = range(1, len(df_dim_product) + 1)
df_dim_product = df_dim_product[['product_key', 'product_id', 'name', 'category']]

# Create dim_support_status with surrogate key
df_dim_support_status = support[['status']].copy()
df_dim_support_status = df_dim_support_status.drop_duplicates(subset=['status'])
df_dim_support_status['support_status_key'] = range(1, len(df_dim_support_status) + 1)
df_dim_support_status = df_dim_support_status[['support_status_key', 'status']]

# Create dim_page with surrogate key
df_dim_page = clickstream[['page_visited']].copy()
df_dim_page = df_dim_page.drop_duplicates(subset=['page_visited'])
df_dim_page['page_key'] = range(1, len(df_dim_page) + 1)
df_dim_page = df_dim_page[['page_key', 'page_visited']]

print("Surrogate keys generated for dimension tables.")
```

### `dim_date`

```python
# Consolidate all relevant dates from fact tables
all_dates = pd.concat([
    customer_orders['order_date'].rename('full_date'),
    pd.to_datetime(clickstream['timestamp']).rename('full_date')
]).drop_duplicates().reset_index(drop=True)

# Ensure all_dates is a Series of datetime objects
all_dates = pd.to_datetime(all_dates)

# Extract unique dates
unique_dates = pd.DataFrame({'full_date': all_dates.dt.normalize().unique()})
unique_dates = unique_dates.sort_values('full_date').reset_index(drop=True)

# Create dim_date
df_dim_date = pd.DataFrame({
    'full_date': pd.to_datetime(unique_dates['full_date'])
})

df_dim_date['date_key'] = (df_dim_date['full_date'].dt.strftime('%Y%m%d')).astype(int)
df_dim_date['year'] = df_dim_date['full_date'].dt.year
df_dim_date['quarter'] = df_dim_date['full_date'].dt.quarter
df_dim_date['month'] = df_dim_date['full_date'].dt.month
df_dim_date['month_name'] = df_dim_date['full_date'].dt.month_name()
df_dim_date['day'] = df_dim_date['full_date'].dt.day
df_dim_date['day_of_week'] = df_dim_date['full_date'].dt.dayofweek # Monday=0, Sunday=6
df_dim_date['day_name'] = df_dim_date['full_date'].dt.day_name()
df_dim_date['week_of_year'] = df_dim_date['full_date'].dt.isocalendar().week.astype(int)
df_dim_date['day_of_year'] = df_dim_date['full_date'].dt.dayofyear
df_dim_date['is_weekend'] = (df_dim_date['full_date'].dt.dayofweek >= 5).astype(bool)
df_dim_date['is_holiday'] = False # Placeholder, would typically join with a holiday calendar

df_dim_date = df_dim_date[[
    'date_key',
    'full_date',
    'year',
    'quarter',
    'month',
    'month_name',
    'day',
    'day_of_week',
    'day_name',
    'week_of_year',
    'day_of_year',
    'is_weekend',
    'is_holiday'
]]

print("Shape of dim_date:", df_dim_date.shape)
print("First 5 rows of dim_date:")
print(df_dim_date.head())
print("\nUnique values in date_key:", df_dim_date['date_key'].nunique())
```

## STEP 6 — Create Fact Tables

Fact tables store the quantitative measures of business processes and foreign keys to link to the dimension tables. We will create `fact_orders`, `fact_support_tickets`, and `fact_clickstream`.

### `fact_orders`

```python
# Merge customer_orders with dim_customer to get customer_key
fact_orders = customer_orders.merge(
    df_dim_customer[['customer_id', 'customer_key']],
    on='customer_id',
    how='left'
)

# Merge with dim_product to get product_key
fact_orders = fact_orders.merge(
    df_dim_product[['product_id', 'product_key']],
    on='product_id',
    how='left'
)

# Merge with dim_date to get date_key
fact_orders = fact_orders.merge(
    df_dim_date[['full_date', 'date_key']],
    left_on='order_date',
    right_on='full_date',
    how='left'
)

# Select and reorder columns for fact_orders
fact_orders = fact_orders[[
    'order_id',
    'customer_key',
    'product_key',
    'date_key',
    'quantity',
    'price',
    'total_amount'
]]

print("Shape of fact_orders:", fact_orders.shape)
print("First 5 rows of fact_orders:")
print(fact_orders.head())
print("\nMissing values in fact_orders:")
print(fact_orders.isnull().sum())
```

### `fact_support_tickets`

```python
# Merge support with dim_customer to get customer_key
fact_support_tickets = support.merge(
    df_dim_customer[['customer_id', 'customer_key']],
    on='customer_id',
    how='left'
)

# Merge with dim_support_status to get support_status_key
fact_support_tickets = fact_support_tickets.merge(
    df_dim_support_status[['status', 'support_status_key']],
    on='status',
    how='left'
)

# Select and reorder columns for fact_support_tickets
fact_support_tickets = fact_support_tickets[[
    'ticket_id',
    'customer_key',
    'support_status_key',
    'issue'
]]

print("Shape of fact_support_tickets:", fact_support_tickets.shape)
print("First 5 rows of fact_support_tickets:")
print(fact_support_tickets.head())
print("\nMissing values in fact_support_tickets:")
print(fact_support_tickets.isnull().sum())
```

### `fact_clickstream`

```python
# Merge clickstream with dim_customer to get customer_key
fact_clickstream = clickstream.merge(
    df_dim_customer[['customer_id', 'customer_key']],
    on='customer_id',
    how='left'
)

# Merge with dim_date to get date_key
fact_clickstream = fact_clickstream.merge(
    df_dim_date[['full_date', 'date_key']],
    left_on=pd.to_datetime(fact_clickstream['timestamp']).dt.normalize(),
    right_on='full_date',
    how='left'
)

# Merge with dim_page to get page_key
fact_clickstream = fact_clickstream.merge(
    df_dim_page[['page_visited', 'page_key']],
    on='page_visited',
    how='left'
)

# Select and reorder columns for fact_clickstream
fact_clickstream = fact_clickstream[[
    'session_id',
    'customer_key',
    'date_key',
    'page_key',
    'timestamp' # Keeping original timestamp for granularity if needed
]]

print("Shape of fact_clickstream:", fact_clickstream.shape)
print("First 5 rows of fact_clickstream:")
print(fact_clickstream.head())
print("\nMissing values in fact_clickstream:")
print(fact_clickstream.isnull().sum())
```

2.  **Check for Duplicates:** Confirm uniqueness constraints on primary keys (surrogate keys) in dimension tables and on business keys/composite keys in fact tables.
3.  **Cross-Table Consistency:** Perform sample queries or checks to ensure data makes logical sense across tables (e.g., total amounts in fact tables align with source data if applicable).
4.  **Data Type Verification:** Re-confirm that data types are appropriate for analysis.

## STEP 11 — Database Loading Preparation (SQL CREATE TABLE statements)

To load our exported data into a relational database, we first need to define the schema using SQL `CREATE TABLE` statements. This section will generate the SQL DDL (Data Definition Language) for each of our dimension and fact tables. We will map pandas data types to appropriate SQL data types.

### Data Type Mapping

*   `int64` -> `INTEGER`
*   `float64` -> `REAL` or `DOUBLE PRECISION`
*   `object` (string) -> `VARCHAR(255)` (or `TEXT` for longer strings)
*   `datetime64[ns]` -> `TIMESTAMP`
*   `bool` -> `BOOLEAN`

```python
def generate_create_table_sql(df, table_name, primary_key=None, foreign_keys={}):
    sql_statements = []
    sql_statements.append(f"CREATE TABLE {table_name} (")

    column_definitions = []
    for col, dtype in df.dtypes.items():
        sql_type = "VARCHAR(255)" # Default for objects
        if pd.api.types.is_integer_dtype(dtype):
            sql_type = "INTEGER"
        elif pd.api.types.is_float_dtype(dtype):
            sql_type = "REAL"
        elif pd.api.types.is_datetime64_any_dtype(dtype):
            sql_type = "TIMESTAMP"
        elif pd.api.types.is_bool_dtype(dtype):
            sql_type = "BOOLEAN"

        definition = f"    {col} {sql_type}"
        if col == primary_key:
            definition += " PRIMARY KEY"
        column_definitions.append(definition)

    # Add foreign key constraints
    for fk_col, ref_table_pk_col in foreign_keys.items():
        ref_table, ref_pk = ref_table_pk_col.split('.')
        column_definitions.append(f"    FOREIGN KEY ({fk_col}) REFERENCES {ref_table}({ref_pk})")

    sql_statements.append(",\n".join(column_definitions))
    sql_statements.append(");")
    return "\n".join(sql_statements)

# Generate SQL for Dimension Tables
sql_dim_customer = generate_create_table_sql(
    df_dim_customer, 'dim_customer', primary_key='customer_key'
)
sql_dim_product = generate_create_table_sql(
    df_dim_product, 'dim_product', primary_key='product_key'
)
sql_dim_date = generate_create_table_sql(
    df_dim_date, 'dim_date', primary_key='date_key'
)
sql_dim_support_status = generate_create_table_sql(
    df_dim_support_status, 'dim_support_status', primary_key='support_status_key'
)
sql_dim_page = generate_create_table_sql(
    df_dim_page, 'dim_page', primary_key='page_key'
)

# Generate SQL for Fact Tables
sql_fact_orders = generate_create_table_sql(
    fact_orders, 'fact_orders', primary_key='order_id',
    foreign_keys={
        'customer_key': 'dim_customer.customer_key',
        'product_key': 'dim_product.product_key',
        'date_key': 'dim_date.date_key'
    }
)
sql_fact_support_tickets = generate_create_table_sql(
    fact_support_tickets, 'fact_support_tickets', primary_key='ticket_id',
    foreign_keys={
        'customer_key': 'dim_customer.customer_key',
        'support_status_key': 'dim_support_status.support_status_key'
    }
)
sql_fact_clickstream = generate_create_table_sql(
    fact_clickstream, 'fact_clickstream', primary_key='session_id',
    foreign_keys={
        'customer_key': 'dim_customer.customer_key',
        'date_key': 'dim_date.date_key',
        'page_key': 'dim_page.page_key'
    }
)

print("--- SQL CREATE TABLE Statements ---")
print("\n" + sql_dim_customer)
print("\n" + sql_dim_product)
print("\n" + sql_dim_date)
print("\n" + sql_dim_support_status)
print("\n" + sql_dim_page)
print("\n" + sql_fact_orders)
print("\n" + sql_fact_support_tickets)
print("\n" + sql_fact_clickstream)
print("\n--- End of SQL Statements ---")
```

## OLAP Cube Simulation with Pandas

To simulate an OLAP cube and demonstrate its analytical capabilities within this environment, we'll use pandas' `pivot_table` functionality. This allows us to perform multi-dimensional analysis by aggregating measures across various dimensions.

We will create a sales cube based on `fact_orders`, linking it to `dim_customer`, `dim_product`, and `dim_date` to explore sales data by customer, product, and time.

```python
import pandas as pd
import os

# Define the path where data warehouse tables were exported
data_warehouse_path = '/content/data_warehouse'

# Ensure the data warehouse directory exists
os.makedirs(data_warehouse_path, exist_ok=True)

# Re-save dimension and fact tables to ensure they are available on disk
# This helps if the directory was lost or previous save operations didn't persist.
# Assumes the DataFrames (dim_customer, dim_product, dim_date, fact_orders) are in memory.
try:
    dim_customer.to_csv(f'{data_warehouse_path}/dim_customer.csv', index=False)
    dim_product.to_csv(f'{data_warehouse_path}/dim_product.csv', index=False)
    dim_date.to_csv(f'{data_warehouse_path}/dim_date.csv', index=False)
    fact_orders.to_csv(f'{data_warehouse_path}/fact_orders.csv', index=False)
    print("Data warehouse tables re-saved to ensure availability for OLAP cube creation.")
except NameError as e:
    print(f"Skipping re-saving data warehouse tables. Some DataFrames were not defined: {e}")
    print("This might happen if previous cells defining these DataFrames were not executed in the current session.")

# Load dimension tables and fact_orders from CSV to ensure they are available for OLAP simulation
# Assuming the files exist or will be created by preceding steps if not.
dim_customer = pd.read_csv(f'{data_warehouse_path}/dim_customer.csv')
dim_product = pd.read_csv(f'{data_warehouse_path}/dim_product.csv')
dim_date = pd.read_csv(f'{data_warehouse_path}/dim_date.csv')
fact_orders = pd.read_csv(f'{data_warehouse_path}/fact_orders.csv')

# Ensure date columns are in datetime format for dim_date if needed for future operations
dim_date['full_date'] = pd.to_datetime(dim_date['full_date'])

# Merge fact_orders with necessary dimension tables to get descriptive attributes
sales_analysis_df = fact_orders.merge(
    dim_customer[['customer_key', 'name', 'gender', 'age']],
    on='customer_key',
    how='left'
).merge(
    dim_product[['product_key', 'name', 'category']],
    on='product_key',
    how='left',
    suffixes=('_customer', '_product')
).merge(
    dim_date[['date_key', 'year', 'month_name', 'day_name']],
    on='date_key',
    how='left'
)

# Display the merged DataFrame head to verify
print("Sales Analysis DataFrame head:")
display(sales_analysis_df.head())

print("\nSales Analysis DataFrame info:")
sales_analysis_df.info()
```

### Creating the Sales OLAP Cube (using `pivot_table`)

Now, let's create our 'cube' to analyze `total_amount` (our measure) by different dimensions.

```python
import pandas as pd

# Cube: Total Sales by Customer and Product Category
sales_by_customer_product_category = pd.pivot_table(
    sales_analysis_df,
    values='total_amount',
    index=['name_customer', 'gender'], # Rows: Customer Name and Gender
    columns=['category'], # Columns: Product Category
    aggfunc='sum',
    fill_value=0
)

print("\nTotal Sales by Customer and Product Category:")
display(sales_by_customer_product_category)

# Cube: Total Sales by Year and Month
sales_by_time = pd.pivot_table(
    sales_analysis_df,
    values='total_amount',
    index=['year'], # Rows: Year
    columns=['month_name'], # Columns: Month Name
    aggfunc='sum',
    fill_value=0
)

print("\nTotal Sales by Year and Month:")
display(sales_by_time)

# Cube: Average Order Value by Customer Age Group
sales_analysis_df['age_group'] = pd.cut(
    sales_analysis_df['age'],
    bins=[0, 25, 35, 45, 60, 100],
    labels=['<25', '25-34', '35-44', '45-59', '60+'],
    right=False
)

avg_order_value_by_age = pd.pivot_table(
    sales_analysis_df,
    values='total_amount',
    index=['age_group'],
    aggfunc='mean',
    fill_value=0
)

print("\nAverage Order Value by Customer Age Group:")
display(avg_order_value_by_age)
```

### OLAP Operations Demonstration

These `pivot_table` operations demonstrate core OLAP capabilities:

*   **Slice:** Filtering data to a specific dimension. (e.g., viewing sales for only 'Product A' category, implicitly done by having it as a column).
*   **Dice:** Selecting a subset of the cube by specifying values for multiple dimensions. (e.g., `sales_by_customer_product_category['Electronics']` to see only electronics sales).
*   **Drill Down/Roll Up:** Changing the level of granularity. (e.g., from `sales_by_time` we can drill down to `day_name` if added to the index/columns, or roll up to `year` which is already shown).

This approach provides a flexible way to explore multi-dimensional data within a pandas DataFrame.
